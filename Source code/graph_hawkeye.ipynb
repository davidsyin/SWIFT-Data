```python
scores = pred_scores
```

```python
test_labels = np.load("turbine_data/t1/test_labels.npy")
# test_labels = np.load("turbine_data/CART2/turbine.z01.b0.20191102.175456.startup.shutdown/test_labels.npy")
# test_labels = np.load("turbine_data/CART2/turbine.z01.b0.20191106.212812.shutdown/test_labels.npy")
```

```python
fprs = []
precs = []
recs = []

for ratio in np.arange(0,100,2):
    thresh = np.percentile(scores, 100 - ratio)
    pred = (scores > thresh).astype(int)
    
    tp = 0 # positive, predicted positive
    tn = 0 # negative, predicted negative
    fp = 0 # negative, predicted positive
    fn = 0 # positive, predicted negative

    for i in range(len(pred)):
        if test_labels[i] == 1:
            if pred[i] == 1:
                tp += 1
            else:
                fn += 1
        else:
            if pred[i] == 1:
                fp += 1
            else:
                tn += 1

    print("Anomaly ratio:", ratio)
    # print("TP:", tp)
    # print("FP:", fp)
    # print("TN:", tn)
    # print("FN:", fn)

    if tp+fp==0 or tp+fn==0:
        continue

    fprs.append(fp/(fp+tn))
    precs.append(tp/(tp+fp))
    recs.append(tp/(tp+fn))
    # print(fprs)
    print("Accuracy:", (tp+tn)/(tp+tn+fp+fn))
    print("Precision: ", tp/(tp+fp))
    print("Recall: ", tp/(tp+fn))
    print("F-score: ", 2*tp/(2*tp+fp+fn))
    print("\n")
```

```python
np.save("results/Hawkeye/fprs_t1.npy", np.array(fprs))
np.save("results/Hawkeye/precs_t1.npy", np.array(precs))
np.save("results/Hawkeye/recs_t1.npy", np.array(recs))
```
