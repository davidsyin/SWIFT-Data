```python
def find_auc(fprs, recs):
    res = 0
    for i in range(1,len(fprs)):
        res += 0.5*(recs[i-1]+recs[i])*abs(fprs[i]-fprs[i-1])
    return res
```

```python
def fscore(precs, recs):
    res = []
    for i in range(len(precs)):
        try:
            res.append(2/(1/precs[i] + 1/recs[i]))
        except:
            pass
    return max(res)
```

```python
def convert_name_wavelets(name):
    name = name.replace("waveletsearch_", "")
    name = name.replace("lvl_1", "1 level").replace("lvl_2", "2 levels").replace("lvl_3", "3 levels").replace("lvl_4", "4 levels").replace("lvl_5", "5 levels").replace("lvl_6", "6 levels")
    name = name.replace("_", ", ")
    return name
```
